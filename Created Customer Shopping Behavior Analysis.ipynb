```python
# creating new column age_group
labels = ['Young', 'Adult', 'Middle-Aged', 'Senior']
# Categorize the age column using pd.cu
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
# Creating column purchase_frequency_days(converting text into number like fotnightly into 14)
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
